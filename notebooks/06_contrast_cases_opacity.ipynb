# 06 Contrast Cases: Opacity Evidence Files

This notebook is the bridge between chart and research claim.

The opacity typology showed a pattern. It did not explain the pattern. Here we choose a small set of contrast cases and start building evidence files for each state.

## Learning Goal

Practice the most important data-research move:

> A chart gives you a question. It does not give you the answer.

We are asking whether public-registry opacity looks more like crisis intensity, administrative practice, privacy policy, source disagreement, or something else.

```python
from pathlib import Path

import pandas as pd

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

CASES_CSV = ROOT / "data" / "processed" / "contrast_cases.csv"
CASE_DIR = ROOT / "docs" / "cases"

cases = pd.read_csv(CASES_CSV)
pd.set_option("display.max_colwidth", 120)
cases[[
    "entidad",
    "role",
    "opacity_typology",
    "total_records",
    "confidential_records",
    "confidential_share",
    "tasa_100k",
    "why_selected",
]]
```

## Compare The Six Cases

These are not rankings. They are contrasts. We want states that help us separate different possible explanations.

```python
comparison_cols = [
    "entidad",
    "opacity_typology",
    "total_records",
    "confidential_records",
    "confidential_share",
    "confidential_records_per100k",
    "tasa_100k",
    "homicide_victims_2015_2025_per100k",
    "kidnapping_victims_2015_2025_per100k",
    "pibe_per_capita_2024_2018_mxn",
    "corruption_prevalence_2025_per100k",
]

cases[comparison_cols].sort_values("confidential_share", ascending=False)
```

## The Evidence Questions

For each state, the generated case file asks:

- What is the confidential share?
- How many records are hidden?
- What public details are missing?
- Does the state have local search/fiscalia rules about data publication?
- Are there official statements about privacy/redaction?
- Do local collectives or journalists criticize registry opacity?
- Has the state had major fosa discoveries or collective searches?
- Does another source disagree with the official pattern?
